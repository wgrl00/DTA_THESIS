Data preparation: extracting URLs for the project subpages that lead directly to the Public Representations tab, which contains the relevant documents.

```python
projects_list = pd.read_csv("scotland.txt", sep ="\t", engine="python")
projects_list.index = projects_list.index + 1
projects_list.head(10)
#list made as of 27.01.2026
```

```python
project_ref = projects_list["ECU Reference"]
project_ref
```

```python
url_base = "https://www.energyconsents.scot/ApplicationDetails.aspx?cr="
url_pubrep = "&T=4"


urls =[]
for item in project_ref:
    target_url = f"{url_base}{item}{url_pubrep}"
    urls.append(target_url)

projects_list["Url"] = urls
```

```python
projects_list.to_csv("energyconsents_df_with_links.csv", index=False)
```
