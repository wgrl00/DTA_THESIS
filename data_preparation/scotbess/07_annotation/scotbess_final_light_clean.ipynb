```python
import re
import unicodedata
import pandas as pd
from google.colab import drive
```

```python

def clean_text(text):
    if not isinstance(text, str):
        return text

    #unicode normalization
    text = unicodedata.normalize("NFKC", text)

    #non-breaking spaces
    text = text.replace("\xa0", " ")

    #common bullet characters
    text = re.sub(
        r"[•▪◦●○■□]+",
        "- ",
        text)

    #repeated separator characters
    text = re.sub(
        r"(?:={3,}|_{3,}|-{4,})",
        " ",
        text)

    # collapse all whitespace/newlines
    text = re.sub(r"\s+", " ", text)

    #remove spaces before punctuation
    text = re.sub(r"\s+([.,;:!?])", r"\1", text)

    #collapse accidental repeated periods
    text = re.sub(r"\.{2,}", ".", text)

    return text.strip()
```

```python
drive.mount("/content/drive")

df = pd.read_csv("/content/drive/MyDrive/thesis_results/SCOTBESS_labeling/v2/SCOTBESS.csv")
```

```python
df["final_masked_text"] = df["final_masked_text"].apply(clean_text)
```

```python
df.to_csv("SCOTBESS_CLEAN.csv", index = False, encoding = "utf-8")
```
