```python
from pathlib import Path
import shutil
import pandas as pd
```

```python
FOLDER = Path("JSON_NO_STRUCTURE_EXTRACTED_TXT")

files = [fp for fp in FOLDER.rglob("*.txt") if "further" in fp.name.lower()]

print(f"Found {len(files)} files\n")

for fp in files:
    print(fp)
```

```python
FOLDER = Path("JSON_NO_STRUCTURE_EXTRACTED_TXT")
DEST = Path("JSON_NO_STRUCTURE_EXTRACTED_TXT_FURTHER_FILES")
DEST.mkdir(exist_ok=True)

files = [fp for fp in FOLDER.rglob("*.txt") if "further" in fp.name.lower()]

for fp in files:
    target = DEST / fp.name
    shutil.move(str(fp), str(target))
```

```python
FOLDER = Path("JSON_NO_STRUCTURE_EXTRACTED_TXT")

rows = []

for fp in FOLDER.rglob("*.txt"):
    try:
        text = fp.read_text(encoding="utf-8")

        rows.append({
            "file": fp.name,
            "chars": len(text),
            "words": len(text.split())})

    except Exception as e:
        print(f"Error reading {fp}: {e}")

df = pd.DataFrame(rows)
print(df.describe())
```

```python
print(df.sort_values("words", ascending=False).head(50))
```

```python
FOLDER = Path("JSON_NO_STRUCTURE_EXTRACTED_TXT")
DEST = Path("JSON_NO_STRUCTURE_EXTRACTED_TXT_OVER_7K_WORDS")
DEST.mkdir(exist_ok=True)

for fp in FOLDER.rglob("*.txt"):
    text = fp.read_text(encoding="utf-8")
    if len(text.split()) > 7000:
        shutil.move(str(fp), str(DEST / fp.name))
```

```python
#inspecting short documents
FOLDER = Path("JSON_NO_STRUCTURE_EXTRACTED_TXT")

rows = []

for fp in FOLDER.rglob("*.txt"):
    try:
        text = fp.read_text(encoding="utf-8").strip()

        rows.append({
            "file_path": str(fp),
            "word_count": len(text.split()),
            "text": text})

    except Exception as e:
        print(f"Error reading {fp}: {e}")

df = pd.DataFrame(rows)

short_df = df[df["word_count"] < 20].copy()

pd.set_option("display.max_colwidth", None)

display(short_df)

len(short_df)
```
