```python
import json
from pathlib import Path
import shutil
```

```python
#handling the representations that have the 'Comment:' section

INPUT_DIR = Path("ALL_PDFS_LPA_EC_JSON")
OUTPUT_DIR = Path("JSON_COMMENT_EXTRACTED")
OUTPUT_DIR.mkdir(exist_ok=True)

def extract_comment_text(blocks):
    collecting = False
    collected = []

    for b in blocks:
        text = b.get("text", "").strip()

        if not text:
            continue

        if not collecting:
            if "Comment Reasons" in text and "Comment:" in text:
                collecting = True
                collected.append(text.split("Comment:", 1)[1].strip())
        else:
            collected.append(text)

    if collected:
        return "\n".join(collected).strip()
    return None

#main loop
json_files = list(INPUT_DIR.rglob("*.json"))

print(f"Found {len(json_files)} JSON files")

count_extracted = 0

for i, json_path in enumerate(json_files, 1):
    try:
        with open(json_path, "r", encoding="utf-8") as f:
            blocks = json.load(f)

        extracted = extract_comment_text(blocks)

        if extracted:
            count_extracted += 1

            #replicate folder structure
            relative_path = json_path.relative_to(INPUT_DIR)
            out_path = OUTPUT_DIR / relative_path.with_suffix(".txt")
            out_path.parent.mkdir(parents=True, exist_ok=True)

            with open(out_path, "w", encoding="utf-8") as f:
                f.write(extracted)

        if i % 100 == 0:
            print(f"{i}/{len(json_files)} processed...")

    except Exception as e:
        print(f"Error processing {json_path}: {e}")


print(f"Extracted Comments section in {count_extracted} files")
```

```python
#moving the files to two folders:
#files containing the 'Comment:' structure 
#remaining files -> I will try to extract the relevant parts from them using a local LLM
INPUT_DIR = Path("ALL_PDFS_LPA_EC_JSON")
OUTPUT_DIR = Path("JSON_WITH_COMMENT_STRUCTURE")
OUTPUT_DIR.mkdir(exist_ok=True)

def has_comment_structure(blocks):
    for b in blocks:
        text = b.get("text", "").strip()
        if "Comment Reasons" in text and "Comment:" in text:
            return True
    return False

json_files = list(INPUT_DIR.rglob("*.json"))
print(f"Found {len(json_files)} JSON files")

count_moved = 0

for i, json_path in enumerate(json_files, 1):
    try:
        with open(json_path, "r", encoding="utf-8") as f:
            blocks = json.load(f)

        if has_comment_structure(blocks):
            relative_path = json_path.relative_to(INPUT_DIR)
            out_path = OUTPUT_DIR / relative_path
            out_path.parent.mkdir(parents=True, exist_ok=True)

            shutil.move(str(json_path), str(out_path))
            count_moved += 1

        if i % 100 == 0:
            print(f"{i}/{len(json_files)} processed...")

    except Exception as e:
        print(f"Error processing {json_path}: {e}")

print(f"Moved {count_moved} JSON files")
```
