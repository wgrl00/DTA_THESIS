```python
import re
import pandas as pd
import unicodedata
```

```python
SCOTBESS = pd.read_csv("SCOTBESS_DATASET_MASKED_REVIEWED_FINAL.csv")
SCOTBESS.head(2)
```

```python
def normalize_final_text(text):
    if pd.isna(text):
        return pd.NA

    text = str(text)

    # uicode normalization
    text = unicodedata.normalize("NFKC", text)

    # standardize line endings
    text = text.replace("\r\n", "\n").replace("\r", "\n")

    # normalize special spaces
    text = (text.replace("\u00a0", " ").replace("\u200b", "").replace("\ufeff", ""))

    # remove control characters, newline and tabs retained
    text = re.sub(r"[\x00-\x08\x0b\x0c\x0e-\x1f\x7f]", "", text)

    #spacing
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r" *\n *", "\n", text)

    #collapsing 3+ line breaks into a single paragraph break
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()
```

```python
SCOTBESS["final_masked_text"] = (SCOTBESS["final_masked_text"].apply(normalize_final_text))
```

```python
SCOTBESS.to_csv("SCOTBESS.csv", index=False, encoding="utf-8")
SCOTBESS.to_excel("SCOTBESS.xlsx", index=False)
```
